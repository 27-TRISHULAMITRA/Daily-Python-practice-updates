```python
df7
```

Above 80 → Excellent, else Average. Create new column called category

```python
def grade(marks):
    if marks>=80:
        return "Excellent"
    elif marks>=50:
        return "Average"
        
        
```

```python
df7['Category']=df7['Average'].apply(grade)
```

```python
df7
```

How will you label Passed column:

True → "Yes", False → "No"

```python
def sort(X):
    if X==True:
        return "Yes"
    elif X==False:
        return "No"
```

```python
df7['Label']=df7['Passed'].apply(sort)
```

```python
df7
```

How will you rename column Math → Maths?

```python
df7.rename(columns={'Math':'Maths'},
           inplace=True)                  #inplace=True, it creates a copy & doesn't update df7
```

```python
df7 #we use dictionary because  a dictionary map old names -> new names 
```

How will you drop column Science?

```python
df7.drop("Science",axis=1)
```

What is the average Math score of class?

```python
print(df7['Maths'].mean())
```
