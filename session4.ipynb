## 4. Try changing an element in tuple. What happens?

A tuple in Python is immutable, which means its elements cannot be changed after creation.
If we try to modify a tuple element, Python will raise an error:
Python
t = (1, 2, 3)
t[0] = 10
👉 This gives:

TypeError: 'tuple' object does not support item assignment
🎯 Conclusion:
Tuples are read-only
Once created, values cannot be modified, added, or removed

## 5. Why should student roll numbers be stored in tuple instead of list?


Student roll numbers should be stored in a tuple instead of a list because:
🔑 1. Immutability (Most Important)
Roll numbers are fixed and should not change
Tuples prevent accidental modification
🔑 2. Data Safety
Since tuples cannot be changed, they ensure data integrity
🔑 3. Better Performance
Tuples are slightly faster and more memory-efficient than lists
🎯 Conclusion:
Tuples are preferred when data should remain constant and protected from changes, such as student roll numbers.
