## break statement ->if breaks the for loop

```python
for i in range(1,11): #here it is telling that when i will be 6 break the for loop
    if i==6: #== is called equality operator
        break
    else:
        print(i)
    
```

## Continue statement-> skip the value 

```python
for i in range(1,11):#here it is telling that when i is 5 skip it & continue 
    if i==5:
        continue 
    else:
        print(i)
```

## pass statement -> python requires at lest one statement  in a block.pass is used to satisfy that requirement  with out adding function.

```python
for i in range(1,11):
    pass
```

```python
for i in range(5): #here the pass does nothing,os the loop runs normally and print all the values from 0 to 4
    if i==3:
        pass #pass is a placeholder 
    print(i)
```

## interview :
 
 1. Difference between break and continue ?

Ans:"break stops the loop completely when a condition is met, while continue skips the current iteration and continues with the next one."

 2. What is pass in Python?
Answer:
"pass is a null statement in Python that does nothing. It is used as a placeholder where a statement is syntactically required but no action needs to be performed."
 
3. Can we write an empty loop in Python?
Answer:
"No, Python does not allow completely empty blocks. However, we can use the pass statement to create an empty loop or block without causing a syntax error."

4. What happens if we don’t use pass in an empty block?
Answer:
"If we leave a block empty without using pass, Python will raise an IndentationError or syntax error because it expects at least one statement inside the block."
