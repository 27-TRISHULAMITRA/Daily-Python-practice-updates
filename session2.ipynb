## prthon doesn't have direct average function so here to get the average we divide sum(of the total of the given by length of that given, here the given thing is test_scores.

Q1. here are your test_scores[45,78,90,66]
Find the total marks,highest,Lower,Average


```python
test_scores=[45,78,90,66]
print(f"Totat marks is{sum(test_scores)}")
print(f"Highest marks is {max(test_scores)}")
print(f"Lowest marks is {min(test_scores)}")
print(f"Average marks is {sum(test_scores)/len(test_scores)}")
```
